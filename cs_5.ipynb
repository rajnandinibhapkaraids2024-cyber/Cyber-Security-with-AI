```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
https://chatgpt.com/share/6ae52e97-3138-83ec-a5bc-079c55ac19e9
```

```python
df=pd.read_csv("04_pwlds_full.csv",skiprows=lambda i:i>0 and np.random.random()>0.01)
```

```python
df
```

```python
df.isnull().sum()
```

```python
dictionary_word=[
    "hello","password","admin","welcome","letmein"]
def extract_feactures(password):
    length=len(password)
    special=int(any(not ch.isalnum() for ch in password))
    dictionary=int(any(word in password.lower()
                       for word in dictionary_word))
    return [length,special,dictionary]

x=pd.DataFrame(
    df["Password"].apply(extract_feactures).tolist(),
    columns=["length","special_character","dictionary_word"]
)

y=df["Strength_Level"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.01,random_state=42)
```

```python
model=RandomForestClassifier(n_estimators=50,random_state=42)
model.fit(x_train,y_train)
predict=model.predict(x_test)
accuracy=accuracy_score(y_test,predict)
print("accuracy:",accuracy)
```

```python
password=input("enter new email:")
feacture=extract_feactures(password)
prediction=model.predict([feacture])

print("prediction:",prediction[0])

if prediction[0]<=1:
    print("password is weak")
elif prediction[0]==2:
    print("password is medium")
else:
    print("password is strong")
```
